# Basic Statistics for the Student Risk Prediction Dataset

This notebook provides an initial exploratory data analysis of the cleaned student dataset. The goal is to summarize the structure of the data and highlight the main distributions that may influence risk prediction modeling.


```python
import pandas as pd
from pathlib import Path

# Load the cleaned dataset
DATA_PATH = Path("../dataset/processed/student_data_cleaned.csv")
df = pd.read_csv(DATA_PATH)

df.head()
```

## 1. Dataset Shape and Data Types

The first step in the analysis is to understand the size of the dataset and the kind of information stored in each column. The shape tells us how many records and features are available, while the column types help determine which variables are categorical or numeric.


```python
print("Dataset shape:")
print(df.shape)

print("\nData types:")
df.dtypes
```

## 2. Missing Values Summary

Missing values can affect model quality and should be reviewed before building predictive systems. This section highlights whether any columns require imputation or cleaning.


```python
missing_values = df.isnull().sum()
missing_values[missing_values > 0]
```

## 3. Descriptive Statistics

Descriptive statistics help summarize the central tendency and spread of the numeric features. These values are useful for understanding the overall behavior of attendance, marks, and backlog patterns in the dataset.


```python
numeric_columns = ["semester", "attendance_pct", "internal_marks", "previous_backlogs"]
df[numeric_columns].describe().T
```

## 4. Department Distribution

Departmental distribution provides insight into whether the student population is balanced across academic units. This is useful for understanding whether the data is representative and whether certain departments show distinct risk patterns.


```python
department_counts = df["department"].value_counts()
department_percentages = df["department"].value_counts(normalize=True) * 100
pd.DataFrame({"count": department_counts, "percentage": department_percentages})
```

## 5. Semester Distribution

The semester distribution helps describe the academic progression of students in the sample. It can reveal whether the dataset is skewed toward early or late semesters and how that may affect risk analysis.


```python
semester_counts = df["semester"].value_counts().sort_index()
semester_percentages = df["semester"].value_counts(normalize=True).sort_index() * 100
pd.DataFrame({"count": semester_counts, "percentage": semester_percentages})
```

## 6. Risk Category Distribution

The final section examines how students are distributed across risk levels. This is a key outcome of the dataset and provides a baseline understanding of the target variable for the risk prediction model.


```python
risk_counts = df["risk_category"].value_counts()
risk_percentages = df["risk_category"].value_counts(normalize=True) * 100
pd.DataFrame({"count": risk_counts, "percentage": risk_percentages})
```
