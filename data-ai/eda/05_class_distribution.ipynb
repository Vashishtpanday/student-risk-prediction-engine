# Class Distribution Analysis for Student Risk Categories

This notebook examines the distribution of the target labels: Low Risk, Moderate Risk, and High Risk. A balanced class distribution is important for reliable model training, while an imbalanced dataset may require special attention such as re-sampling or class weighting.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load cleaned dataset
DATA_PATH = Path("../dataset/processed/student_data_cleaned.csv")
df = pd.read_csv(DATA_PATH)

plt.style.use("seaborn-v0_8-whitegrid")
sns.set_theme(style="whitegrid")

value_counts = df["risk_category"].value_counts()
value_counts
```

## Pie Chart

The pie chart provides a simple visual comparison of the proportions of Low Risk, Moderate Risk, and High Risk students in the dataset. It is useful for quickly assessing whether one class dominates the sample.


```python
plt.figure(figsize=(7, 7))
plt.pie(
    value_counts.values,
    labels=value_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=["#4C78A8", "#F58518", "#54A24B"],
)
plt.title("Risk Category Distribution")
plt.axis("equal")
plt.tight_layout()
plt.show()
```

## Count Plot

The count plot shows the absolute number of students in each risk category. This complements the pie chart by making the class sizes easy to compare directly.


```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="risk_category", order=["Low Risk", "Moderate Risk", "High Risk"], palette="deep")
plt.title("Risk Category Count Plot")
plt.xlabel("Risk Category")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

## Percentage Table

The percentage table turns the counts into proportions, which makes it easier to judge whether the dataset is balanced or whether one class is overrepresented.


```python
percentage_table = (value_counts / len(df) * 100).round(2)
percentage_table
```

## Interpretation

The class balance is assessed by comparing the percentages of each risk category. If one class is much smaller than the others, the dataset is considered imbalanced and may require resampling or weighting during model training.
