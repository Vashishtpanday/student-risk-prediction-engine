# Univariate Analysis of Student Risk Factors

This notebook explores the distribution of individual variables in the cleaned student dataset. Univariate analysis helps highlight patterns in attendance, academic performance, department composition, and risk-related categories before moving to more advanced modeling.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load cleaned dataset
DATA_PATH = Path("../dataset/processed/student_data_cleaned.csv")
df = pd.read_csv(DATA_PATH)

# Style configuration
plt.style.use("seaborn-v0_8-whitegrid")
sns.set_theme(style="whitegrid")

df.head()
```

## Attendance Distribution

The attendance histogram shows how frequently students fall into different attendance ranges. This is one of the strongest indicators of academic risk because low attendance often correlates with weaker performance and increased dropout risk.


```python
plt.figure(figsize=(8, 5))
sns.histplot(df["attendance_pct"], bins=20, kde=True, color="steelblue")
plt.title("Attendance Percentage Distribution")
plt.xlabel("Attendance (%)")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

## Internal Marks Distribution

The internal marks histogram shows the spread of academic performance scores. This view helps identify whether the class is generally performing well or whether a large number of students are struggling academically.


```python
plt.figure(figsize=(8, 5))
sns.histplot(df["internal_marks"], bins=20, kde=True, color="mediumseagreen")
plt.title("Internal Marks Distribution")
plt.xlabel("Internal Marks")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

## Semester Distribution

The semester distribution shows how many students are present in each academic year. This helps assess whether the dataset is balanced across semesters or concentrated in a smaller subset of the student population.


```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="semester", palette="Set2")
plt.title("Semester Distribution")
plt.xlabel("Semester")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

## Department Distribution

The department count plot reveals how students are distributed across academic departments. This can help identify whether the dataset is balanced or whether certain departments dominate the sample.


```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="department", palette="pastel")
plt.title("Department Count Distribution")
plt.xlabel("Department")
plt.ylabel("Count")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## Risk Category Distribution

The risk category count plot summarizes the target labels in the dataset. It is important to understand whether the data is balanced across risk classes or whether some categories dominate the sample.


```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="risk_category", palette="deep")
plt.title("Risk Category Distribution")
plt.xlabel("Risk Category")
plt.ylabel("Count")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## CP/NCP Distribution

The CP/NCP count plot indicates whether students are more often associated with good or poor academic standing based on their course progress indicator. This helps interpret the relationship between academic outcomes and the available categorical feature.


```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="cp_ncp", palette="rocket")
plt.title("CP/NCP Distribution")
plt.xlabel("CP/NCP")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

## Previous Backlogs Distribution

The previous backlog distribution highlights the number of backlogs carried by students. This is useful because backlog history is often a strong signal of recurring academic difficulties and future risk.


```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="previous_backlogs", palette="viridis")
plt.title("Previous Backlogs Distribution")
plt.xlabel("Previous Backlogs")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
