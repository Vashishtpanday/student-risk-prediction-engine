# Correlation Analysis for Student Risk Prediction

This notebook evaluates the relationship between numerical features in the cleaned student dataset and the target risk behavior. Correlation analysis helps identify which variables are most strongly associated with academic risk and should therefore receive more attention during model development.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load cleaned dataset
DATA_PATH = Path("../dataset/processed/student_data_cleaned.csv")
df = pd.read_csv(DATA_PATH)

# Prepare a numeric dataset for correlation analysis
numeric_df = df[["semester", "attendance_pct", "internal_marks", "previous_backlogs"]].copy()

plt.style.use("seaborn-v0_8-whitegrid")
sns.set_theme(style="whitegrid")

numeric_df.head()
```

## Correlation Matrix

The correlation matrix quantifies the strength and direction of linear relationships among the numeric variables. This helps identify which academic indicators move together and which variables may provide the most useful signal for predicting student risk.


```python
correlation_matrix = numeric_df.corr()
correlation_matrix
```

## Correlation Heatmap

The heatmap visually highlights the strength of correlations between the numerical variables. Darker shades indicate stronger relationships, making it easier to spot which features are moving together and which ones may be especially useful for risk prediction.


```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

## Pairplot

A pairplot displays scatterplots and density distributions for each pair of variables. This helps uncover non-linear patterns and possible clusters that may not be obvious from the correlation matrix alone.


```python
sns.pairplot(numeric_df, diag_kind="kde")
plt.suptitle("Pairplot of Numerical Features", y=1.02)
plt.show()
```

## Correlation Summary

The correlation summary helps interpret which features are most likely to influence academic risk. In practice, attendance, internal marks, and backlog history are expected to be the strongest indicators because they directly reflect student engagement and academic performance.


```python
# Rank the features by absolute correlation with the most relevant academic performance variables
correlation_summary = correlation_matrix.abs().stack().reset_index()
correlation_summary.columns = ["feature_1", "feature_2", "correlation"]
correlation_summary = correlation_summary[correlation_summary["feature_1"] != correlation_summary["feature_2"]]
correlation_summary.sort_values("correlation", ascending=False).head(10)
```
