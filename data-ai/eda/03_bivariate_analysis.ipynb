# Bivariate Analysis of Student Risk Factors

This notebook explores relationships between individual student features and the target risk category. Bivariate analysis highlights whether variables such as attendance, marks, semester, department, and backlog history are meaningfully associated with different levels of academic risk.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load cleaned dataset
DATA_PATH = Path("../dataset/processed/student_data_cleaned.csv")
df = pd.read_csv(DATA_PATH)

plt.style.use("seaborn-v0_8-whitegrid")
sns.set_theme(style="whitegrid")

df.head()
```

## Attendance vs Risk Category

This boxplot compares attendance distributions across the different risk categories. Higher attendance is expected to be associated with lower risk, while lower attendance should appear more frequently in high-risk students.


```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="risk_category", y="attendance_pct", palette="Set2")
plt.title("Attendance vs Risk Category")
plt.xlabel("Risk Category")
plt.ylabel("Attendance (%)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## Internal Marks vs Risk Category

This violin plot shows how internal marks vary across risk levels. Stronger students should generally appear in lower-risk groups, while weaker performers should be concentrated in higher-risk categories.


```python
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x="risk_category", y="internal_marks", palette="pastel")
plt.title("Internal Marks vs Risk Category")
plt.xlabel("Risk Category")
plt.ylabel("Internal Marks")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## Semester vs Risk Category

This grouped bar chart compares the distribution of semesters within each risk category. It helps reveal whether students in certain academic stages are more likely to fall into high-risk or low-risk groups.


```python
semester_risk = pd.crosstab(df["semester"], df["risk_category"])
plt.figure(figsize=(8, 5))
semester_risk.plot(kind="bar", figsize=(10, 5), color=["#4C78A8", "#F58518", "#54A24B"])
plt.title("Semester vs Risk Category")
plt.xlabel("Semester")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Department vs Risk Category

This grouped bar chart highlights how risk categories are distributed across departments. It can reveal whether certain departments have a higher share of high-risk students and whether the risk profile differs by academic unit.


```python
department_risk = pd.crosstab(df["department"], df["risk_category"])
plt.figure(figsize=(8, 5))
department_risk.plot(kind="bar", figsize=(10, 5), color=["#4C78A8", "#F58518", "#54A24B"])
plt.title("Department vs Risk Category")
plt.xlabel("Department")
plt.ylabel("Count")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## Backlogs vs Risk Category

This grouped bar chart displays the relationship between backlog history and risk level. A higher number of previous backlogs often points to recurring academic difficulty and should be associated with elevated student risk.


```python
backlog_risk = pd.crosstab(df["previous_backlogs"], df["risk_category"])
plt.figure(figsize=(8, 5))
backlog_risk.plot(kind="bar", figsize=(10, 5), color=["#4C78A8", "#F58518", "#54A24B"])
plt.title("Backlogs vs Risk Category")
plt.xlabel("Previous Backlogs")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
